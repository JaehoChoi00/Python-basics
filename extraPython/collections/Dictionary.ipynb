```python
dictionaryExample = {"Key 1" : "Value 1", "Key 2" : "Value 2"}
print(dictionaryExample)
```

```python
print(dictionaryExample["Key 1"])

print(dictionaryExample.get("Key 1"))

print(dictionaryExample.get("Key 3"))

print(dictionaryExample.get("Key 3", "There is nothing"))

dictionaryExampleYes = {"Replacement Key 1" : "Replacement Value 1"}

print(dictionaryExample.get("Key 3", dictionaryExampleYes))
```

```python
dictionaryExample2 = {"Key 1" : "Value 1", "Key 2" : "Error"}
print(dictionaryExample2)

dictionaryExample2["Key 2"] = "Value 2"
print(dictionaryExample2)
```

```python
dictionaryExample3 = {"Key 1" : "Value 1", "Key 2" : "Value 2", "Unwanted key 3" : "Unwanted Value 1", "Unwanted key 4" : "Unwanted Value 4"}
print(dictionaryExample3)

dictionaryExample3.pop("Unwanted key 3")
print(dictionaryExample3)

dictionaryExample3.popitem()
print(dictionaryExample3)
```

```python
for keys in dictionaryExample2:
    print(keys)

for keys in dictionaryExample2:
    print(dictionaryExample2[keys])

for key, value in dictionaryExample2.items():
    print(f"{key}: {value}")
```

```python
nestedDictionaryExample1 = {"Parent Key 1" : {"Child Key 1" : "Child Value 1"}, 
                            "Parent Key 2" : {"Child Key 2" : "Child Value 2"}
                            }
print(nestedDictionaryExample1)
```

```python
print(nestedDictionaryExample1.get("Parent Key 1"))

print(nestedDictionaryExample1.get("Parent Key 1").get("Child Key 1"))

print(nestedDictionaryExample1.get("Parent Key 1").get("Child Key 2"))

print(nestedDictionaryExample1.get("Parent Key 2").get("Child Key 2"))

print(nestedDictionaryExample1.get("Parent Key 3", {}).get("Child Key 2"))
```
