## Set Notation Functions

```python
fruitBasketA = set(("Apple", "Banana", "Cherry", "Orange", "Melon"))
fruitBasketB = set(("Apple", "Banana", "Cherry", "Orange", "Lemon", "Blueberry"))
fruitBasketC = set(("Apple", "Banana", "Cherry", "Orange"))
fruitBasketD = set(("Apple", "Banana", "Cherry", "Mango"))

print("A " + str(fruitBasketA))
print("B " + str(fruitBasketB))
print("C " + str(fruitBasketC))

print("A - B = " + str(fruitBasketA.difference(fruitBasketB)))
print("B - A = " + str(fruitBasketB.difference(fruitBasketA)))
print("A ∪ B = " + str(fruitBasketA.union(fruitBasketB)))
print("A ∩ B = " + str(fruitBasketA.intersection(fruitBasketB)))
print("C ⊂ A = " + str(fruitBasketC.issubset(fruitBasketA)))
print("C ⊂ B = " + str(fruitBasketC.issubset(fruitBasketB)))
print("C ⊂ D = " + str(fruitBasketC.issubset(fruitBasketD)))
```
