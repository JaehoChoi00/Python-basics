```python
from abc import ABC, abstractmethod
```

```python
class ClassABC(ABC):

    @abstractmethod
    def __init__(self, valueToAssign):
        self.assignedValue = valueToAssign

    @abstractmethod
    def methodOne(self):
        pass

    @property
    @abstractmethod
    def methodTwo(self):
        pass

class ClassExtendingAbstract(ClassABC):
    def __init__(self, valueToAssign):
        super().__init__(valueToAssign)
        
    def methodOne(self):
        return "Regular Abstract Method"
    
    @property
    def methodTwo(self):
        return "@property enforces value two"
```

```python
valueToAssign = "HELLO!"

instance = ClassExtendingAbstract(valueToAssign)

print("Assigned Value: " + str(instance.assignedValue))
print("Method One: " + instance.methodOne())
print("Method Two: " + instance.methodTwo)
```
