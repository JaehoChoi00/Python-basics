```python
class Person:

    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

    def changeName(self, newName):
        if (self.name != newName):
            self.name = newName
        else:
            print("Name is the same as before")

    def growUp(self):
        self.age += 1

    def printStats(self):
        print(f"Person name: {self.name} \nPerson age: {self.age} \nPerson gender: {self.gender}")
        print(f"________________________________________________________________________________")

people = set()

person1 = Person("Bob", 15, "Male")
person2 = Person("Alex", 14, "Female")

people.add(person1)
people.add(person2)

yearsToPass = 5

for person in people:
    person.printStats()

for years in range (yearsToPass):
    for person in people:
        person.growUp()

for person in people:
    person.printStats()
```

Interestingly. If you run the code above multiple times. You will see that the order of output changes. Sometimes its Bob first, and sometimes Alex.

And the interesting fact is that Everytime we restart the program, the hashing of variables change internally. And that yields in not the same order.

```python
class Point2D:

    def __init__(self, x, y):
        self._x = x
        self._y = y

    @property
    def x(self):
        return self._x

    @property
    def y(self):
        return self._y

    def __eq__(self, other):
        if not isinstance(other, Point2D):
            return False
        return self._x == other._x and self._y == other._y

    def __hash__(self):
        return hash((self._x, self._y))

    def __repr__(self):
        return f"Point({self._x}, {self._y})"

visitedLocations = set()

point_a = Point2D(10, 20)
point_b = Point2D(5, 5)
point_c = Point2D(10, 20)

visitedLocations.add(point_a)
visitedLocations.add(point_b)
visitedLocations.add(point_c)  

print("Unique visited points:", visitedLocations)

check_point = Point2D(5, 5)
if check_point in visitedLocations:
    print("Player has been to (5, 5) before!")
```

```python
class Student:
    # A class-level attribute
    school_name = "Python Academy"

    def __init__(self, name):
        self.name = name  # 'self' refers to the unique instance

    @classmethod
    def get_school(cls):
        # 'cls' refers to the Student class itself
        return cls.school_name


student1 = Student("Bob")

print(student1.get_school())
```
